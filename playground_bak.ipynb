```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path


file_numbers = [(1, .25), (2, .5), (3, .75),(4, 1.) ]
frames = []

for nbr, alpha in file_numbers:
    path = Path(f"data/superoperator_N4_h_sweep_{nbr}.npz")  # adjust filename
    data = np.load(path)
    
    df = pd.DataFrame({
        "h": data["h"],
        "Delta2": data["Delta2"],
        "Delta_th": data["Delta_th"],
        "num_closer": data["num_closer"],
        "trace_distance": data["trace_distance"],
        "get_mixingtime": data["get_mixingtime"],
    })
    df["alpha"] = alpha
    frames.append(df)

ha_data = pd.concat(frames, ignore_index=True)

print("columns:", ha_data.columns.tolist())
print(ha_data.head())
```

```python
mix_grid = ha_data.pivot(index="h", columns="alpha", values="get_mixingtime").sort_index().sort_index(axis=1)
gap_grid = ha_data.pivot(index="h", columns="alpha", values="Delta2").sort_index().sort_index(axis=1)
tr_grid = ha_data.pivot(index="h", columns="alpha", values="trace_distance").sort_index().sort_index(axis=1)

h_vals = mix_grid.index.to_numpy()
alpha_vals = mix_grid.columns.to_numpy()
values = ["mixingtime", "Delta2", "trace_distance"]

Y = np.stack([
    mix_grid.to_numpy(),
    gap_grid.to_numpy(),
    tr_grid.to_numpy(),
], axis=1)



for alpha_idx in range(0,4):
    x = h_vals
    y0 = Y[:,0,alpha_idx]
    plt.plot(x,y0, label=rf"$\alpha = {alpha_vals[alpha_idx]}$", marker=".")
    plt.legend(loc="upper center")
    plt.xlabel("h / J")
    plt.ylabel("iterations")
    plt.title(rf"Convergence to channel-fixedpoint $(\epsilon = 0.05)$")
    plt.grid(True)
    plt.savefig("docu/iterations.png", dpi=300, bbox_inches="tight")
```

```python
ax = sns.barplot(data=ha_data, x="h", y="Delta2", hue="alpha", dodge=True, palette="tab10")

ticks = ax.get_xticks()
labels = [f"{v:.1f}" for v in sorted(ha_data["h"].unique())]

ax.set_xticks(ticks[1::2])
ax.set_xticklabels(labels[1::2])

ax.set_xlabel("h / J")
ax.set_ylabel(rf"$\Delta_2$")

ax.set_title("Spectral gap of averaged channel")
plt.legend(title=rf"$\alpha$")
ax.figure.savefig("docu/gap.png", dpi=300, bbox_inches="tight")
```
